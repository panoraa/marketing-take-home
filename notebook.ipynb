```python
import pandas as pd, numpy as np

meta = df2.drop(columns=["date_dt"], errors="ignore").copy()
adj  = df1.drop(columns=["date_dt"], errors="ignore").copy()
adj_raw_rows = len(adj)

for df in (meta, adj):
    df["date"] = pd.to_datetime(df["date"], format="%m/%d/%y")

START, END = pd.Timestamp("2025-06-01"), pd.Timestamp("2025-08-31")
MATURE_CUTOFF = END - pd.Timedelta(days=30)   # 2025-08-01

dq_log = []
def log(source, issue, rows, action):
    dq_log.append({"source": source, "issue": issue, "rows": int(rows), "action": action})
    print(f"LOGGED  [{source}] {issue}: {int(rows):,} rows -> {action}")

print("meta:", meta.shape, "| adj:", adj.shape)
```

```python
def split_meta(col):
    parts = col.str.split(" | ", n=1, regex=False)
    return parts.str[0].str.strip(), parts.str[1].str.strip()

for lvl in ["campaign", "adset", "ad"]:
    meta[f"{lvl}_id"], meta[f"{lvl}_name"] = split_meta(meta[lvl])
meta["network"] = "Meta"

print(meta[["campaign_id", "campaign_name"]].drop_duplicates().to_string(index=False))
assert meta["campaign_id"].str.isdigit().all(), "Some campaign IDs are not numbers"
print("\nCHECK OK: every ID is numeric ->", meta.campaign_id.nunique(), "campaigns")
```

```python
n = (meta["country"] == "U.S.").sum()
meta["country"] = meta["country"].replace({"U.S.": "US"})
log("Meta", "Country 'U.S.'", n, "Standardised to US")

unk = meta["country"] == "unknown"
geo = meta["campaign_name"].str.extract(r"^(?:MT-)?([A-Z]{2})[-_]")[0]
log("Meta", "Country 'unknown'", unk.sum(),
    f"Filled from campaign name (${meta.loc[unk, 'spend'].sum():.2f} spend)")
meta.loc[unk, "country"] = geo[unk]

meta["market"] = meta["country"].map({"US": "United States", "CA": "Canada", "AU": "Australia"})
print(meta["market"].value_counts())
assert meta["market"].notna().all(), "Some rows have no market"
print("CHECK OK: only 3 markets, none missing")
```

```python
KEYS = ["date", "network", "market", "os", "campaign_id"]
m_agg = meta.groupby(KEYS, as_index=False)[["spend", "clicks", "impressions"]].sum()
a_agg = adj.groupby(KEYS, as_index=False)[["installs", "purchases",
                                           "d0_purchase_revenue", "d30_purchase_revenue"]].sum()
merged = m_agg.merge(a_agg, on=KEYS, how="outer")

names = pd.concat([meta[["campaign_id", "campaign_name"]], adj[["campaign_id", "campaign_name"]]])
merged["campaign_name"] = merged["campaign_id"].map(
    names.drop_duplicates("campaign_id").set_index("campaign_id")["campaign_name"])
for c in ["installs", "purchases", "d0_purchase_revenue", "d30_purchase_revenue"]:
    merged[c] = merged[c].fillna(0)
is_m = merged.network.eq("Meta")
for c in ["spend", "clicks", "impressions"]:
    merged.loc[is_m, c] = merged.loc[is_m, c].fillna(0)   # Google cost stays blank
merged["d30_mature"] = merged["date"] <= MATURE_CUTOFF
merged["month"] = merged["date"].dt.strftime("%Y-%m")

log("Merged", "Installs after Aug 1 lack full D30 window", (merged.date > MATURE_CUTOFF).sum(),
    "d30_mature flag; D30 ROAS on mature cohorts only")

merged.to_csv("cleaned_merged.csv", index=False)
print(merged.shape)
```

```python
def kpis(df, by):
    s = df.groupby(by)[["spend", "clicks", "impressions", "installs", "purchases",
                        "d0_purchase_revenue", "d30_purchase_revenue"]].sum()
    mat = df[df.d30_mature].groupby(by)[["spend", "d30_purchase_revenue"]].sum()
    nz = lambda x: x.replace(0, np.nan)
    s["CPI"] = s.spend / nz(s.installs)
    s["CTR"] = s.clicks / nz(s.impressions)
    s["D0_ROAS"] = s.d0_purchase_revenue / nz(s.spend)
    s["D30_ROAS_mature"] = mat.d30_purchase_revenue / nz(mat.spend)
    return s

M = merged[merged.network == "Meta"].assign(total="Total")
cols = ["spend", "installs", "CPI", "CTR", "D0_ROAS", "D30_ROAS_mature"]
for by in ["total", "market", "os", "month", "campaign_name"]:
    display(kpis(M, by)[cols].sort_values("spend", ascending=False).round(3))
```

```python
def kpis(df, by):
    s = df.groupby(by)[["spend", "clicks", "impressions", "installs", "purchases",
                        "d0_purchase_revenue", "d30_purchase_revenue"]].sum()
    mat = df[df.d30_mature].groupby(by)[["spend", "installs", "d30_purchase_revenue"]].sum()
    nz = lambda x: x.replace(0, np.nan)
    s["CPM"]  = s.spend / nz(s.impressions) * 1000
    s["CTR"]  = s.clicks / nz(s.impressions)
    s["CPC"]  = s.spend / nz(s.clicks)
    s["IPM"]  = s.installs / nz(s.impressions) * 1000
    s["CVR_click_install"] = s.installs / nz(s.clicks)
    s["CPI"]  = s.spend / nz(s.installs)
    s["purch_per_install"] = s.purchases / nz(s.installs)
    s["cost_per_purchase"] = s.spend / nz(s.purchases)
    s["rev_per_purchase_D0"] = s.d0_purchase_revenue / nz(s.purchases)
    s["D0_rev_per_install"]  = s.d0_purchase_revenue / nz(s.installs)
    s["D30_rev_per_install_mature"] = mat.d30_purchase_revenue / nz(mat.installs)
    s["D0_ROAS"] = s.d0_purchase_revenue / nz(s.spend)
    s["D30_ROAS_mature"] = mat.d30_purchase_revenue / nz(mat.spend)
    s["D30_D0_multiplier"] = s.D30_ROAS_mature / s.D0_ROAS
    s["payback"] = np.select([s.D0_ROAS >= 1, s.D30_ROAS_mature >= 1], ["Day 0", "< 30 days"], "> 30 days")
    s.loc[s.spend.isna() | (s.spend == 0), "payback"] = "n/a"
    return s

M = merged[merged.network == "Meta"].assign(total="Meta total")
CORE = ["spend", "installs", "CPI", "CTR", "D0_ROAS", "D30_ROAS_mature", "payback"]
display(kpis(M, "total").T)
```

```python
def split_meta(col):
    p = col.str.split(" | ", n=1, regex=False)
    return p.str[0].str.strip(), p.str[1].str.strip()
for lvl in ["campaign", "adset", "ad"]:
    meta[f"{lvl}_id"], meta[f"{lvl}_name"] = split_meta(meta[lvl])
meta["network"] = "Meta"

n = (meta.country == "U.S.").sum()
meta["country"] = meta.country.replace({"U.S.": "US"})
log("Meta", "Country 'U.S.'", n, "Standardised to US")
unk = meta.country == "unknown"
geo = meta.campaign_name.str.extract(r"^(?:MT-)?([A-Z]{2})[-_]")[0]
log("Meta", "Country 'unknown'", unk.sum(), f"Filled from campaign name (${meta.loc[unk,'spend'].sum():.2f} spend)")
meta.loc[unk, "country"] = geo[unk]
meta["market"] = meta.country.map({"US": "United States", "CA": "Canada", "AU": "Australia"})

u = meta.campaign_name.str.upper()
ios, andr = u.str.contains("IOS"), u.str.contains(r"(?:^|[-_ ])AND(?:ROID)?(?:[-_ ]|$)")
meta["os_name"] = np.select([ios & andr, ios, andr], ["conflict", "iOS", "Android"], "unknown")
assert meta.market.notna().all()
print(meta.market.value_counts().to_dict(), "|", meta.os_name.value_counts().to_dict())
```

```python
n = adj.duplicated().sum(); adj = adj.drop_duplicates()
log("Adjust", "Exact duplicate rows", n, "Removed")
exp = adj.campaign.eq("Expired Attributions")
log("Adjust", "'Expired Attributions', no campaign ID", exp.sum(), f"Removed ({adj.loc[exp,'installs'].sum()} installs)")
adj = adj[~exp].copy()
n = (adj.os == "android-tv").sum()
adj["os"] = adj.os.replace({"android-tv": "android"}).map({"android": "Android", "ios": "iOS"})
log("Adjust", "OS 'android-tv'", n, "Grouped into Android")

adj["network"] = adj.partner.map({"facebook": "Meta", "adwords": "Google"})
def split_adjust(col):
    return (col.str.extract(r"\((\d+)\)\s*$")[0],
            col.str.replace(r"\s*\(\d+\)\s*$", "", regex=True).str.strip())
for lvl in ["campaign", "adset", "ad"]:
    adj[f"{lvl}_id"], adj[f"{lvl}_name"] = split_adjust(adj[lvl])

log("Adjust", "spend, clicks, impressions = 0 in all rows", len(adj_src), "Meta cost from Meta file; Google cost N/A")
log("Adjust", "purchase_revenue = d0_purchase_revenue in all rows", len(adj_src), "Not used; asked client")
adj["refund_flag"] = adj.d30_purchase_revenue < adj.d0_purchase_revenue
log("Adjust", "D30 revenue below D0", adj.refund_flag.sum(), "Kept, flagged as possible refunds")
assert adj.campaign_id.notna().all()
print("Adjust rows:", len(adj))
```

```python
camp_market = meta.drop_duplicates("campaign_id").set_index("campaign_id")["market"]
is_meta = adj.network.eq("Meta")
adj["device_country"] = adj.country
adj["market"] = adj.country
adj.loc[is_meta, "market"] = adj.loc[is_meta, "campaign_id"].map(camp_market)
outside = is_meta & adj.market.ne(adj.device_country)
log("Adjust", "Meta installs outside campaign target market", outside.sum(),
    f"Assigned to campaign market ({adj.loc[outside,'installs'].sum()} installs)")

adj_os = (adj[is_meta].groupby(["campaign_id", "os"]).installs.sum().reset_index()
          .sort_values("installs").drop_duplicates("campaign_id", keep="last")
          .set_index("campaign_id")["os"])
bad = meta.os_name.isin(["conflict", "unknown"])
log("Meta", "Campaign name contains both IOS and AND", bad.sum(), "OS taken from Adjust for same campaign ID")
meta["os"] = np.where(bad, meta.campaign_id.map(adj_os), meta.os_name)

log("Adjust", "Rows before Jun 1 (Jan only; Feb-May missing)", (adj.date < START).sum(), "Excluded from comparison window")
adj  = adj[adj.date.between(START, END)].copy()
meta = meta[meta.date.between(START, END)].copy()
assert adj.installs.sum() == 293600
print("OK. Installs Jun-Aug:", adj.installs.sum())
```

```python
KEYS = ["date", "network", "market", "os", "campaign_id"]
m_agg = meta.groupby(KEYS, as_index=False)[["spend", "clicks", "impressions"]].sum()
a_agg = adj.groupby(KEYS, as_index=False)[["installs", "purchases", "d0_purchase_revenue", "d30_purchase_revenue"]].sum()
merged = m_agg.merge(a_agg, on=KEYS, how="outer")

names = pd.concat([meta[["campaign_id", "campaign_name"]], adj[["campaign_id", "campaign_name"]]])
merged["campaign_name"] = merged.campaign_id.map(names.drop_duplicates("campaign_id").set_index("campaign_id")["campaign_name"])
for c in ["installs", "purchases", "d0_purchase_revenue", "d30_purchase_revenue"]:
    merged[c] = merged[c].fillna(0)
is_m = merged.network.eq("Meta")
for c in ["spend", "clicks", "impressions"]:
    merged.loc[is_m, c] = merged.loc[is_m, c].fillna(0)      # Google cost stays blank
merged["d30_mature"] = merged.date <= MATURE_CUTOFF
merged["month"] = merged.date.dt.strftime("%Y-%m")
merged["week"] = merged.date.dt.to_period("W-SUN").dt.start_time
log("Merged", "Installs after Aug 1 lack full D30 window", (merged.date > MATURE_CUTOFF).sum(), "D30 ROAS on mature cohorts only")
merged.to_csv("cleaned_merged.csv", index=False)
print(merged.shape)
```

```python
def kpis(df, by):
    s = df.groupby(by)[["spend", "clicks", "impressions", "installs", "purchases",
                        "d0_purchase_revenue", "d30_purchase_revenue"]].sum()
    mat = df[df.d30_mature].groupby(by)[["spend", "installs", "d30_purchase_revenue"]].sum()
    nz = lambda x: x.replace(0, np.nan)
    s["CPM"]  = s.spend / nz(s.impressions) * 1000
    s["CTR"]  = s.clicks / nz(s.impressions)
    s["CPC"]  = s.spend / nz(s.clicks)
    s["IPM"]  = s.installs / nz(s.impressions) * 1000
    s["CVR_click_install"] = s.installs / nz(s.clicks)
    s["CPI"]  = s.spend / nz(s.installs)
    s["purch_per_install"] = s.purchases / nz(s.installs)
    s["cost_per_purchase"] = s.spend / nz(s.purchases)
    s["rev_per_purchase_D0"] = s.d0_purchase_revenue / nz(s.purchases)
    s["D0_rev_per_install"]  = s.d0_purchase_revenue / nz(s.installs)
    s["D30_rev_per_install_mature"] = mat.d30_purchase_revenue / nz(mat.installs)
    s["D0_ROAS"] = s.d0_purchase_revenue / nz(s.spend)
    s["D30_ROAS_mature"] = mat.d30_purchase_revenue / nz(mat.spend)
    s["D30_D0_multiplier"] = s.D30_ROAS_mature / s.D0_ROAS
    s["payback"] = np.select([s.D0_ROAS >= 1, s.D30_ROAS_mature >= 1], ["Day 0", "< 30 days"], "> 30 days")
    s.loc[s.spend.isna() | (s.spend == 0), "payback"] = "n/a"
    return s

M = merged[merged.network == "Meta"].assign(total="Meta total")
CORE = ["spend", "installs", "CPI", "CTR", "D0_ROAS", "D30_ROAS_mature", "payback"]
display(kpis(M, "total").T)
```

```python
cj = kpis(M[M.month == "2025-07"], "campaign_name")[["spend", "installs", "CPI"]]
ca = kpis(M[M.month == "2025-08"], "campaign_name")[["spend", "installs", "CPI"]]
ca = ca[ca.spend > 0]
cpi_jul = cj.spend.sum() / cj.installs.sum()
cpi_aug = ca.spend.sum() / ca.installs.sum()

# Counterfactual: August budget split, but every continuing campaign at its July CPI
cont = ca.index.intersection(cj[cj.installs > 0].index)
cf_installs = ca.installs.copy()
cf_installs[cont] = ca.loc[cont, "spend"] / cj.loc[cont, "CPI"]
cpi_cf = ca.spend.sum() / cf_installs.sum()

print(f"CPI July ${cpi_jul:.2f} -> August ${cpi_aug:.2f}  (+${cpi_aug - cpi_jul:.2f})")
print(f"  Mix effect  (budget moved to new/pricier campaigns): +${cpi_cf - cpi_jul:.2f}")
print(f"  Rate effect (same campaigns got more expensive):     +${cpi_aug - cpi_cf:.2f}")
new = ca.index.difference(cont)
print(f"\nNew campaigns in August: {len(new)}, ${ca.loc[new,'spend'].sum():,.0f} spend at CPI ${ca.loc[new,'spend'].sum()/ca.loc[new,'installs'].sum():.2f}")
display(pd.DataFrame({"CPI_Jul": cj.CPI, "CPI_Aug": ca.CPI, "spend_Aug": ca.spend}).loc[ca.index].sort_values("spend_Aug", ascending=False).round(2))
```
